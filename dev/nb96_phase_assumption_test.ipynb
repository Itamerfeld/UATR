```python

path = '/Users/itamarmerfeld/Projects/UATR/data/croatia/2307_freediving/RBW6737_20250723_101100.wav'
data, fs = librosa.load(path)
F, T, Sxx, phasogram = ut.calc_spectrogram(data, fs=fs, nperseg=nperseg, percent_overlap=overlap, window=window, remove_dc=dc, crop_freq=crop_freq)

fig = go.Figure()
fig.add_trace(go.Heatmap(x=T, y=F, z=Sxx, colorscale='Viridis', showscale=False))
fig.update_layout(width=1200, height=800)
fig.show()
```
